```python
#  x array from -10 to 10 with step 0.001
x = np.arange(-10, 10, 0.001)

# calculate y-values
y = mx.lab_1_problem_1(x, student_id)

# Plot and decorate
plt.figure(figsize=(9, 6))
plt.plot(x, y, label="f(x)")
plt.axhline(y=1, color="r", linestyle="--", label="y = 1")
plt.title(f"Function f(x) for student {student_id}")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.legend()
plt.show()
```

## Problem 2

```python
# x and y are already known.
# Find the indices where the function crosses y = 1
idx = np.where(np.diff(np.sign(y - 1)))[0]
approx = [round(float(x[i]), 1) for i in idx]
print("Approximate crossing positions:", approx)

# One zoomed graph for each intersection point
for n, x_c in enumerate(approx, start=1):
    plt.figure(figsize=(7, 5))
    plt.plot(x, y, label="f(x)")
    plt.xlim(x_c - 0.5, x_c + 0.5)   # zoom x-axis around the crossing
    plt.ylim(0, 2)                   # keeps y = 1 in view
    plt.axhline(y=1, color="r", linestyle="--", label="y = 1")
    plt.xlabel("x")
    plt.ylabel("f(x)")
    plt.title(f"Point {n} - Zoomed view")
    plt.grid(True)
    plt.legend()
    plt.show()

# Check values (compare with the graphs)
for n, i in enumerate(idx, start=1):
    print(f"Point {n}: x is approximately {x[i]:.2f}")
```
