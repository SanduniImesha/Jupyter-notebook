```python
k = 3.8
c = 5.2
x = np.linspace(0, 10, 5000)
f = np.sqrt(k * x)
g = (x - c) ** 2

# overview graph
plt.figure()
plt.plot(x, f, label='f(x)')
plt.plot(x, g, label='g(x)')
plt.ylim(0, 10)
plt.title('Problem 2 - overview')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()

# intersection 1
plt.figure()
plt.plot(x, f, label='f(x)')
plt.plot(x, g, label='g(x)')
plt.xlim(3.2, 3.4)
plt.ylim(3.3, 3.9)
plt.title('Intersection 1 (zoomed)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()

# intersection 2
plt.figure()
plt.plot(x, f, label='f(x)')
plt.plot(x, g, label='g(x)')
plt.xlim(7.4, 7.6)
plt.ylim(4.6, 5.4)
plt.title('Intersection 2 (zoomed)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```
