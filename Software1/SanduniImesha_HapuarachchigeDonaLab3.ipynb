# Problem 1

```python
txt = mx.lab_3_problem_1(student_id)
display(Markdown(txt))
```

### My solution

```python
x = np.linspace(-10, 10, 2001)      # step 0.01, smooth graph

# f(x) from the problem text, one boolean mask for each region
y = ((x < -0.8) * 0 +
     (x >= -0.8) * (x < 2.2) * (0.8 + x) +
     (x >= 2.2) * (x < 5.2) * (5.2 - x) +
     (x >= 5.2) * 0)

# check the values at the boundaries (should be continuous)
x_test = np.array([-0.8, 2.2, 5.2])
y_test = ((x_test < -0.8) * 0 +
          (x_test >= -0.8) * (x_test < 2.2) * (0.8 + x_test) +
          (x_test >= 2.2) * (x_test < 5.2) * (5.2 - x_test) +
          (x_test >= 5.2) * 0)
print(x_test, y_test)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label='f(x)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Problem 1: Piecewise Function')
plt.legend()
plt.grid(True)
plt.show()
```

## Problem 2


```python
txt = mx.lab_3_problem_2(student_id)
display(Markdown(txt))
```

### My solution

```python
x = np.linspace(-10, 10, 2001)      # same step 0.01 as in problem 1

y = ((x < -7.1) * -0.46 +
     (x >= -7.1) * (x < 4.5) * (0.1 * x**3 + 1.45 * x**2 + 6.77 * x + 10.3) +
     (x >= 4.5) * 79.24)

x_test = np.array([-7.11, -7.1, -7.09, 4.49, 4.5, 4.51])
y_test = ((x_test < -7.1) * -0.46 +
          (x_test >= -7.1) * (x_test < 4.5) * (0.1 * x_test**3 + 1.45 * x_test**2 + 6.77 * x_test + 10.3) +
          (x_test >= 4.5) * 79.24)
print(x_test, np.round(y_test, 2))

plt.figure(figsize=(8, 5))
plt.plot(x, y, label='f(x)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Problem 2: Piecewise Function')
plt.legend()
plt.grid(True)
plt.show()
```
